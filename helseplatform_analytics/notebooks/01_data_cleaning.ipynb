```python
import pandas as pd
import numpy as np
import os
```

```python
population_data_county=pd.read_excel(set_path + 'rawdata/07459_ Population_county.xlsx', skiprows=4, nrows=424,header=None)
population_data_county.columns=['County', 'Sex', 'Age'] + list(range(2015, 2027))
population_data_county[["County", "Sex"]] = population_data_county[ ["County", "Sex"]].ffill()
population_data_county.Sex=population_data_county.Sex.replace({'Females': 'F', 'Males': 'M'})
population_data_county["County_code"] = population_data_county["County"].str.extract(r'(F-\d+)')
population_data_county["County_name"] = population_data_county["County"].str.replace(r'(F-\d+)\s','',regex=True)
population_data_county["Age_num"] = (population_data_county["Age"].str.extract(r'(\d+)'))
population_data_county["Age_num"] = population_data_county["Age_num"].astype(int)
population_data_county = population_data_county.melt(id_vars=["County","Sex","Age", 'Age_num', 'County_code', 'County_name'],var_name="Year",value_name="Population")
population_data_county.Year=population_data_county.Year.astype(int)
population_data_county = population_data_county[population_data_county["Year"] <= 2025]
population_data_county["County_name"] = (population_data_county["County_name"] .replace({ "Trøndelag - Trööndelage": "Trøndelag" }))
population_data_county.head()
population_data_county.to_csv(set_path + 'cleandata/population_county_clean.csv',  encoding="utf-8-sig",index=False)
```

```python
population_data_municipality=pd.read_excel(set_path + 'rawdata/07459_population_municipalities.xlsx', skiprows=4, nrows=12936-4,header=None)
population_data_municipality.columns=['Municipality', 'Sex', 'Age'] + list(range(2015, 2026))
population_data_municipality[["Municipality", "Sex"]] = population_data_municipality[ ["Municipality", "Sex"]].ffill()
population_data_municipality.Sex=population_data_municipality.Sex.replace({'Females': 'F', 'Males': 'M'})
population_data_municipality["Municipality_code"] = population_data_municipality["Municipality"].str.extract(r'(K-\d+)')
population_data_municipality["Municipality_name"] = population_data_municipality["Municipality"].str.replace(r'(K-\d+)\s','',regex=True)
population_data_municipality["Age_num"] = (population_data_municipality["Age"].str.extract(r'(\d+)'))
population_data_municipality["Age_num"] = population_data_municipality["Age_num"].astype(int)
population_data_municipality = population_data_municipality.melt(id_vars=["Municipality","Sex","Age", 'Age_num', 'Municipality_code', 'Municipality_name'],var_name="Year",value_name="Population")
population_data_municipality.Year=population_data_municipality.Year.astype(int)
population_data_municipality.head()
population_data_municipality.to_csv(set_path +'cleandata/population_municipality_clean.csv', encoding="utf-8-sig",index=False)
```

```python
physician_data=pd.read_excel(set_path + 'rawdata/11996_Man-years for physicians.xlsx', skiprows=4, nrows=1801, header=None)
physician_data.columns=['Wage_Agreement', 'Function', 'Municipality'] + list(range(2015, 2026))
physician_data[["Wage_Agreement", "Function", "Municipality"]] = physician_data[ ["Wage_Agreement", "Function", "Municipality"]].ffill()
physician_data["Municipality_code"] = physician_data["Municipality"].str.extract(r'(\d+)')
physician_data["Municipality_name"] = physician_data["Municipality"].str.replace(r'(\d+)\s','',regex=True)
physician_data = physician_data.melt(id_vars=["Wage_Agreement","Function","Municipality",  'Municipality_code', 'Municipality_name'],value_vars=list(range(2015, 2026)), var_name="Year", value_name="Physician_man_years")
physician_data.Year=physician_data.Year.astype(int)
physician_data["Physician_man_years"] = pd.to_numeric(physician_data["Physician_man_years"],errors="coerce")
physician_data = physician_data[physician_data["Municipality_code"].notna()]
physician_data["Municipality_code"] = ("K-" + physician_data["Municipality_code"].astype(str))
physician_data.head()
physician_data.tail(10)
physician_data.info()
physician_data.to_csv(set_path +'cleandata/physician_clean_data.csv',  encoding="utf-8-sig",index=False)
```

```python
nurse_data=pd.read_excel(set_path + 'rawdata/14533_Man-years for nurse.xlsx', skiprows=4, nrows=720, header=None)
nurse_data.columns=['Municipality', 'Occupation'] + list(range(2015, 2026))
nurse_data[["Municipality", "Occupation"]] = nurse_data[ ["Municipality", "Occupation"]].ffill()
nurse_data["Municipality_code"] = nurse_data["Municipality"].str.extract(r'(\d+)')
nurse_data["Municipality_name"] = nurse_data["Municipality"].str.replace(r'(\d+)\s','',regex=True)
nurse_data = nurse_data.melt(id_vars=["Municipality","Occupation", 'Municipality_code', 'Municipality_name'],var_name="Year",value_name="Nurse_man_years")
nurse_data.Year=nurse_data.Year.astype(int)
nurse_data["Nurse_man_years"] = pd.to_numeric(nurse_data["Nurse_man_years"],errors="coerce")
nurse_data = nurse_data[nurse_data["Municipality_code"].notna()]
nurse_data["Municipality_code"] = ("K-" + nurse_data["Municipality_code"].astype(str))
nurse_data.to_csv(set_path +'cleandata/nurse_clean_data.csv',  encoding="utf-8", index=False)
```
